# 3-Data Processing

```python
data=data.drop(['ID','Doors'],axis=1)
```

```python
data.head()
```

# Date

```python
import datetime
dtime=datetime.datetime.now()
```

```python
#هطرح تاريخ سنه الانتاج بتاعت العربيه  من تاريخ انهارده عشان اعرف عمر العربيه
data['Age']=dtime.year-data['Prod. year']
```

```python
data=data.drop(['Prod. year'],axis=1) #هنا مسحت السنه لان مبقاش ليها لازمه بعد م طلعت عمر العربيه
```

# Levy

```python
data.Levy.replace({'-':'0'},inplace=True)
data['Levy']=data['Levy'].astype(int)
```

# Mileage

```python
data.columns
```

```python
data['Mileage']=data['Mileage'].str.replace('km',' ')
```

# Engine Volume

```python
data['Engine volume'].unique()
```

```python
data['Engine volume']=data['Engine volume'].str.replace('Turbo',' ')
data['Engine volume']=data['Engine volume'].astype(float)
```

```python
data.info()
```

# Detect OutLier

```python
data_numeric=data.select_dtypes(exclude='object')
for col in data_numeric:
    q1=data[col].quantile(0.25)
    q3=data[col].quantile(0.75)
    iqr=q3-q1
    low=q1-1.5*iqr
    high=q3+1.5*iqr
    outlier=((data_numeric[col]>high)|(data_numeric[col]<low)).sum()
    total=data_numeric[col].shape[0]
    print(f"Total Outliers in {col} are :{outlier}--{round(100*(outlier)/total,2)}%")
    if outlier>0:
        data=data.loc[(data[col]<=high) & (data[col]>=low)]
```

# Transfrom Data

```python
dobject=data.select_dtypes(include='object')
dnumeric=data.select_dtypes(exclude='object')
```

```python
la=LabelEncoder()
```

```python
for i in range(0,dobject.shape[1]):
    dobject.iloc[:,i]=la.fit_transform(dobject.iloc[:,i])
```

```python
dobject = dobject.astype('int64')
```

```python
data=pd.concat([dobject,dnumeric],axis=1)
```

```python
data.info()
```

# 4-Model

```python
x=data.drop('Price',axis=1)
y=data['Price']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
Algorithm=['LinearRegression','DecisionTreeRegressor','RandomForestRegressor',
           'GradientBoostingRegressor','XGBRFRegressor','SVR']
R2=[]
RMSE=[]
```

```python
def models(model):
    model.fit(x_train,y_train)
    pre=model.predict(x_test)
    r2=r2_score(y_test,pre)
    rmse=np.sqrt(mean_squared_error(y_test,pre))
    R2.append(r2)
    RMSE.append(rmse)
    score=model.score(x_test,y_test)
    print(f"The Score of Model is :{score}")
    
```

```python
model1=LinearRegression()
model2=DecisionTreeRegressor()
model3=RandomForestRegressor()
model4=GradientBoostingRegressor()
model5=XGBRFRegressor()
model6=SVR()
```

```python
models(model1)
models(model2)
models(model3)
models(model4)
models(model5)
models(model6)
```

```python
df=pd.DataFrame({'Algorithm':Algorithm,'R2_Score':R2,'RMSE':RMSE})
df
```

```python
fig,sx=plt.subplots(figsize=(20,5))
plt.plot(df.Algorithm,df.R2_Score,label='R2_Score',marker='v')
plt.legend()
plt.show()
```

```python
fig,sx=plt.subplots(figsize=(20,5))
plt.plot(df.Algorithm,df.RMSE,label='RMSE',color='r',marker='o')
plt.legend()
plt.show()
```
